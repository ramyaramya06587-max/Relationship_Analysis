```python
# Task 29: Relationship Analysis
# Veda Technology - AI & ML Internship

import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Dataset
df = pd.read_csv("iris.csv")

print("Dataset Preview:")
print(df.head())

# 2. Select Two Numerical Columns
x = df["sepal_length"]
y = df["petal_length"]

# 3. Create Scatter Plot
plt.figure(figsize=(8, 5))

plt.scatter(x, y, color="blue", marker="o")

plt.title("Relationship Between Sepal Length and Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.grid(True)

plt.savefig("scatter_plot.png")
plt.show()

# 4. Calculate Correlation
correlation = x.corr(y)

print("\nCorrelation:", correlation)

# 5. Describe Relationship
if correlation > 0.5:
    print("There is a positive relationship.")
elif correlation < -0.5:
    print("There is a negative relationship.")
else:
    print("There is a weak relationship.")

# 6. Observations
print("\nObservations:")
print("1. The scatter plot shows the relationship between two numerical variables.")
print("2. Sepal length and petal length generally increase together.")
print("3. The variables show a positive relationship.")
```
